```python
import pandas as pd
df = pd.read_csv('flight_price.csv')
df.head()
```

```python
df.info()
```

```python
X = df.drop(['id','flight','price'], axis=1)
y = df['price']
```

## Cuando una variables es categórica tengo 2 opciones
* ORDINAL
* NOMINAL
- Para la regresión lineal es mejor usar dummies
### DUMMIES
* Es una categoría menos (de 8 variables se conviertene en 7)
* Tomar una categoría de referencia (arbitrariamente)
*


```python
X = pd.get_dummies(X, drop_first=True, dtype=int)  #drop_first -> el primero que sale es el que toma como referencia   || Aerolinea es la referencia en el ejemplo
X
```

```python
import statsmodels.api as sm

X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())
```

ELIMINAR VARIABLES
* Más de 0.05 se elimina UNA POR UNA

INTERPRETAR COEFICIENTES
* airline_GO_FIRST CUESTA (CAMBIA) 1621.8750 MÁS QUE LA REFERENCIA
* Salir de source_city_Chennai  CUESTA -67.4818 (MENOS) QUE LA REFERENCIA
* La clase class_Economy CUESTA -4.492e+04  (MENOS) QUE LA REFERENCIA (BUSSINES)
* Cada hora que dure tu vuelo (duration) CUESTA  42.5731 MÁS

```python
X = X.drop(['destination_city_Mumbai'], axis=1)
model = sm.OLS(y, X).fit()
print(model.summary())
```

ELIMINAR UNA DUMMIE ES QUE NO HAY DIFERENCIA DE ESA VARIABLE CON LA DE REFERENCIA

```python
X = X.drop(['source_city_Chennai'], axis=1)
model = sm.OLS(y, X).fit()
print(model.summary())
```

```python
X = X.drop(['airline_Air_India'], axis=1)
model = sm.OLS(y, X).fit()
print(model.summary())
```
