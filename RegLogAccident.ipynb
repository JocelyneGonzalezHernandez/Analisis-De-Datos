```python
#Importar bibliotecas
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("accident.csv")
df
```

PARSIMONIA

```python
df.gender = df.gender.map({"Male": 1, "Female":0 })
df.helmet = df.helmet.map({"Yes": 1, "No": 0})
df.seatbelt = df.seatbelt.map({"Yes": 1, "No": 0})

df
```

```python
from sklearn.model_selection import train_test_split
x = df.drop("survived", axis=1)
y = df.survived

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=31)

from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

logit_model = smf.logit('survived ~ age + speed + helmet + seatbelt + gender', data=df)
result = logit_model.fit()
print(result.summary())
```

```python
#QUITAR SEATBEALT Y GENERO
import statsmodels.api as sm
import statsmodels.formula.api as smf

df_cleaned = df.dropna()

logit_model = smf.logit('survived ~ age + speed + helmet', data=df_cleaned)
result = logit_model.fit()
print(result.summary())
```

```python
#Orden de importancia de las variables
# A mayor valor, más importancia
from sklearn.inspection import permutation_importance
import matplotlib.pyplot as plt

result=permutation_importance(model, x_train, y_train, n_repeats=10, random_state=42, n_jobs=-1)
print(result.importances_mean)
```
